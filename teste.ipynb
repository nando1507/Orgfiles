```python
from pathlib import Path
from shutil import move, rmtree
from tqdm import tqdm
from os import rmdir
```

```python
pastas = list(Path('C:/Reddit/pessoas_organizadas').glob("Pessoa*"))
```

```python
for pasta in tqdm(pastas, total=len(pastas)):
    arquivos = list(pasta.rglob("*"))
    if len(arquivos) < 10:
        for arquivo in tqdm(arquivos, total=len(arquivos)):
            if not "foto_referencia" in arquivo.name: 
                # print(arquivo)
                move(arquivo, Path("C:/Reddit") / arquivo.name)
        # break
    if len(arquivos)  == 1:
        for arquivo in tqdm(arquivos, total=len(arquivos)):
            if "foto_referencia" in arquivo.name: 
                rmtree(pasta)
                # pasta.rmdir()
```
